```python

def factorial(num):
    if isinstance (num,int)==False:
        return "not a number"
    if num<0:
        return "negative number"
    elif num==0:
        return 1
    elif num==1:
        return 1
    else:
        return num*factorial(num-1)

def test_factorial():
    try:
        count=0
        temp=factorial(2)
        assert temp==2, "failed for positive number"
        count+=1
        print("passed for positive number")
        temp2=factorial(-4)
        assert temp2=="negative number", "failed for negative numbers"
        count+=1
        print("passed for negative numbers")
        temp3=factorial(0)
        assert temp3==1, "failed for 0"
        count+=1
        print("passed for 0")
        temp4=factorial("s")
        assert temp4=="not a number", "failed for string"
        count+=1
        print("passed for string")
    except AssertionError as e:
        print(e)
        
    finally:
        print("total test casese passed",count)
    
test_factorial()
```

```python
def factorial(num):
    if isinstance (num,int)==False:
        return "not a integer"
    if num<0:
        return "negative number"
    elif num==0 or num==1:
        return 1
    else:
        return num*factorial(num-1)

def test_factorial():
    try:
        count=0
        temp=factorial(2)
        assert temp==2, "failed for positive number"
        count+=1
        print("passed for positive number")
    except AssertionError as e:
        print(e)
    try:
        temp2=factorial(-4)
        assert temp2=="negative number", "failed for negative numbers"
        count+=1
        print("passed for negative numbers")
    except AssertionError as e:
        print(e)
    try:
        temp3=factorial(0)
        assert temp3==1, "failed for 0"
        count+=1
        print("passed for 0")
    except AssertionError as e:
        print(e)
    try:
        temp4=factorial("s")
        assert temp4=="not a integer", "failed for string"
        count+=1
        print("passed for string")
    except AssertionError as e:
        print(e)
        
    finally:
        print("total test casese passed",count)
    
test_factorial()
```

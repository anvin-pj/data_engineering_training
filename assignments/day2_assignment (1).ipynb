```python
import reduce from functools
#assignment 1
#why we cannot call local varible from outside
#the local varible has only the inside the function .if we use it outside the function it shows the name error

#assignment 2
#where to use map,reduce,filter
#map
#it is used to map or modify each element by a function
num1=[1,2,3,4]
lst1=list(map(lambda x:x+2,num1))
print(lst1)

#filter
#it is used to filter out only needed things using a condition
num2=[1,-2,3,4,0]
lst2=list(filter(lambda x:x>0,num2))
print(lst2)

#reduce
#it is used to reduce the list into a single value as it gave the total product of a list or sum
num3=[1,-2,3,4,5]
lst3=list(reduce(lambda x,y:x*y,num3))
print(lst3)
```

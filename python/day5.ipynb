```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
months=['jan','feb','mar','apr','may','jun']
traffic=[150,2,180,220,250,310]
plt.plot(months,traffic)
plt.show()
```

```python
months=['jan','feb','mar','apr','may','jun']
traffic=[150,2,180,220,250,310]
plt.plot(months,traffic,marker='^',markersize=10,color='red',linestyle='-.',linewidth=3)
plt.xlabel("Month")
plt.ylabel("Monthly Traffic")
plt.title("Monthly Website Traffic")
plt.show()
```

```python
months=['jan','feb','mar','apr','may','jun']
pa=[150,2,180,220,250,310]
pb=[100,30,50,270,150,150]

#create a line plot for product A with blue line and circular markers
plt.plot(months,pa,marker='o',linestyle='-',color='blue',label='product A')

#create a line plot for product B with red line and square markers
plt.plot(months,pb,marker='s',linestyle='-',color='red',label='product B')

#add
plt.xlabel("Month")
plt.ylabel("Monthly revenue")
plt.title("Monthly Revenue comparision")

#diplay a legend to differniate between product A and product B
plt.legend()

#display the plot
plt.show()
```

```python
overs=['(0-5)','(5-10)','(10-15)','(15-20)']
teama=[30,70,120,200]
teamb=[40,60,130,180]

#create a line plot for product A with blue line and circular markers
plt.plot(overs,teama,marker='o',linestyle='-.',color='blue',label='TEAM A')

#create a line plot for product B with red line and square markers
plt.plot(overs,teamb,marker='s',linestyle='-.',color='red',label='TEAM B')

#add
plt.xlabel("OVERS")
plt.ylabel("TEAM TOTAL")
plt.title("MATCH A")

#diplay a legend to differniate between product A and product B
plt.legend()

#display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

categories=['category A','category B','category C','category D']
values=[20,35,30,25]

plt.bar(categories,values,color=['red','green','black','violet'],width=0.8)

plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar plot Example")
plt.show()
```

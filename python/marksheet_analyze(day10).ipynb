```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv(r"/Users/Administrator/downloads/student_marksheet(new).csv")

# Question columns
question_cols = df.columns[1:]

# Total score
df["Total_Score"] = df[question_cols].sum(axis=1)

# Percentage
total_questions = len(question_cols)
df["Percentage"] = (df["Total_Score"] / total_questions) * 100
d=df.groupby('Percentage')['Total_Score'].size()
d
```

```python
# Classification
def classify(student):
    if student >= 70:
        return "Advanced"

    else:
        return "Basic"

df["Category"] = df["Percentage"].apply(classify)
df
```

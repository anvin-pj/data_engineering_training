```python
def factorial(num=" "):
    if isinstance (num,float):
        if int(num)==num:
            num=int(num)
    if num>9999:
        raise valueError
    if num<0:
        raise ValueError
    elif num==0 or num==1:
        return 1
    else:
        return num*factorial(num-1)
```

```python
import unittest
class TestFactorial(unittest.TestCase):
    
    def test_positive(self):
        self.assertEqual(factorial(6), 720)

    def test_zero(self):
        self.assertEqual(factorial(0), 1)

    def test_negative(self):
        with self.assertRaises(ValueError):
            factorial(-4)

    def test_string(self):
        with self.assertRaises(TypeError):
            factorial("a")

unittest.main(argv=[''], verbosity=2, exit=False)
```

```python
#generator
#A generator is a special type of iterator that is defined using a func and the yeild keyword
#generate value 1 at atime
def my_gen():
    for i in range(1,6):
        yield i
gen=my_gen()

for value in (gen):
    print(value)
```

```python
def my_gen():
    for i in range(1,6):
        yield i
gen=my_gen()

for index,value in enumerate(gen):
    print(index,value)
```

```python
#diff between iterator and generator
#iterator requires class with __iter__() and __next__()
#generator: uses function with yield keyword
#iterator more code,manual implementation
#generator simple and consise implementaion
#iterator:slightly more control
```

```python
def odd(num1,num2):
    for i in range(num1,num2+1):
        if i%2!=0:
            yield i
gen=odd(10,20)
for i in gen:
    print(i)
gen=odd(20,30)
for i in gen:
    print(i)
```

```python
class show_cart:
    def __iter__(self):
        self.cart=["dress","bag","shoes"]
        self.i=0
        return self
    def __next__(self):
        if self.i<len(self.cart):
            item= self.cart[self.i]
            self.i+=1
            return item
        else:
            raise StopIteration
myclass=show_cart()
m=iter(myclass)
for i in m:
    print(i)

            
```
